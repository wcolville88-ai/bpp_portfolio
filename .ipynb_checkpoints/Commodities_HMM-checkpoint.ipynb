# Global Commodity Analysis with Hidden Markov Models
#### A Hidden Markov Model (HMM) is a probabilistic model for time series data where observed values are driven by unobserved “hidden” states. At each time point, the system is assumed to be in one of several latent states, each governing the behaviour of the observed process. HMMs are widely used in areas such as finance, speech recognition, and bioinformatics to model systems with regime-like structure that cannot be directly observed.

#### In financial contexts, HMMs are useful for capturing changes in market regimes. Commodity prices often move through distinct phases, such as stable trends, periods of high volatility, or sudden shocks driven by macroeconomic or geopolitical events. Rather than assuming a single generating process, an HMM allows these different behaviours to be represented explicitly through separate latent states.

#### This project applies an HMM to global commodity price data from the World Bank’s Pink Sheet, a long-running and standardised monthly dataset covering energy, metal, and agricultural commodities. Its consistency and broad coverage make it suitable for analysing long-term structural dynamics across markets.

#### The goal is to infer hidden market states that drive observed price movements, and to examine how commodities transition between regimes over time. These states are interpreted in terms of trends, volatility, and shock-driven periods, providing a structured way to understand market behaviour beyond raw price movements.

#### Overall, the model offers a probabilistic framework for identifying and interpreting underlying commodity market regimes in a transparent and data-driven way.

### Preparing data for the HMM
#### Converting the data to log-prices and log-difference of prices.
#### - This is the standard in financial modelling and improves stability of the HMM.

```python
# Deriving log-prices and adding them to the dataframe

price = {}
price_cols = []

for commodity in commodities:
    series = pd.Series(np.log(df[commodity]), name = f"{commodity}_price")
    price[f"{commodity}_price"] = series
    price_cols.append(f"{commodity}_price")

prices_df = pd.concat(price.values(), axis = 1)
df = pd.concat([df, prices_df], axis = 1)
```

```python
# Deriving 1 month commodity log(returns) and adding them to the dataframe

returns = {}
returns_cols = []

for commodity in commodities:
    series = pd.Series(np.log(df[commodity]) - np.log(df[commodity]).shift(periods = 1), name = f"{commodity}_return")
    returns[f"{commodity}_return"] = series
    returns_cols.append(f"{commodity}_return")

returns_df = pd.concat(returns.values(), axis = 1)
df = pd.concat([df, returns_df], axis = 1)
```

### Feature Engineering
#### Features were engineered to capture 3 unique signals:
#### - Pricing trend -  Is the price going up or down and how fast? Derived by fitting a linear line over a rolling 6 month window and taking the gradient. 
#### - Returns volatility - Are returns stable or jumping up and down erratically? Derived by taking the mean absolute deviation over a rolling 6 month window. MAD is less sensitive to outliers.
#### - Shock - Has the price suddenly shifted up or down? Derived by taking the absolute returns for the period.

```python
# Commodity Trend

commodity_trend = {}
commodity_trend_cols = []

for commodity in df[price_cols]:
    series = pd.Series(
        df[commodity].rolling(window = 6).apply(lambda x: np.polyfit(range(0, len(x)), x, 1)[0], raw = True) , name = f"{commodity}_trend")
    commodity_trend[f"{commodity}_trend"] = series
    commodity_trend_cols.append(f"{commodity}_trend")

commodity_trend_df = pd.concat(commodity_trend.values(), axis = 1)
df = pd.concat([df, commodity_trend_df], axis = 1)
```

```python
# Commodity Volatility

commodity_volatility = {}
commodity_volatility_cols =[]

for commodity in df[returns_cols]:
    series = pd.Series(df[commodity].rolling(window = 6, min_periods = 1).apply(lambda x: np.mean(np.abs(x - np.mean(x))), raw = True), name = f"{commodity}_volatility")
    commodity_volatility[f"{commodity}_volatility"] = series
    commodity_volatility_cols.append(f"{commodity}_volatility")

commodity_volatility_df = pd.concat(commodity_volatility.values(), axis = 1)
df = pd.concat([df, commodity_volatility_df], axis = 1).sort_values(by = "date", ascending = True).reset_index(drop = True)
```

```python
# Commodity shock

commodity_shock = {}
commodity_shock_cols = []

for commodity in df[returns_cols]:
    series = pd.Series(df[commodity].abs(), name = f"{commodity}_shock")
    commodity_shock[f"{commodity}_shock"] = series
    commodity_shock_cols.append(f"{commodity}_shock")

commodity_shock_df = pd.concat(commodity_shock.values(), axis = 1)
df = pd.concat([df, commodity_shock_df], axis = 1).sort_values(by = "date", ascending = True).reset_index(drop = True)
```

```python
df = df[df["date"] >= "1990-01-01"].reset_index(drop = True)
```

### Checking for multicollinearity

#### Feature sets should be free from multicollinearity. Multicollinearity can lead to unstable models and difficulty separating distinct states. This was tested via condition number and should be kept under 30. No commodity failed this test.

```python
# Checking for multicollinearity

data_checks = []

for commodity in commodities:
    results = [
        commodity
        , np.linalg.cond(np.cov(df[[f"{commodity}_price_trend", f"{commodity}_return_volatility", f"{commodity}_return_shock"]].values, rowvar=False))]
    data_checks.append(results)

df_data_checks = pd.DataFrame(data_checks, columns = ["commodity", "feature_set_condition_number"]).sort_values(by = "feature_set_condition_number", ascending = False).reset_index(drop = True)
df_data_checks.head()
```

### Choosing a commodity
##### Given the current uncertainty with regards to the price of oil (due to the Iran war 2026) **crude oil, brent** was chosen to take forward for analysis although the methods applied should be sound for any of the commodities.

```python
# Getting relevant commodity columns

commodity = "crude oil, brent"
commodity_cols = [x for x in df if x.startswith(f"{commodity}") and x.endswith(("trend", "volatility", "shock"))]
```

### Splitting the data

#### For a purely descriptive project, splitting the data isn't strictly necessary however it is a good test of whether the model is over fit.
#### An 80/20 temporal split was applied before the data was standardised, a requirement to ensure a stable model for long sequences.

```python
# Split into training and test datasets
series_length = len(df)
training_rows = int(series_length * 0.8)
train = df[commodity_cols].copy()[:training_rows].to_numpy()
test = df[commodity_cols].copy()[training_rows:].to_numpy()

scaler = StandardScaler()
X_train = scaler.fit_transform(train)
X_test = scaler.transform(test)
```

### Finding the best model - Hyperparameter tuning
#### The GaussianHMM model from hmmlearn was chosen because:
#### - The data was continuous.
#### - It is typically in financial contexts to assume emission distributions, within states, are typically Gaussian in nature - that is - approximately normally distributed.
#####
#### We are going to loop through:
#### - 50 different random states - HMMs (when initial parameters are not specified) start with random parameters. Testing hyperparameters over multiple random states assessed model stability across random initialisations.
#### - n_components - We'll try 2 to 6 state models.
#### - covariance_type - Essentially setting how emissions (observed data) might interact and vary within states.
#####
#### Model performance was assessed via:
#### - Bayesian Information Criterion (BIC) - a metric that balances model performance (log-likelihood) with model complexity (chosen parameters) - an effort to prevent overfitting.
#### - The standard deviation of BIC over all random states for the particular hyperparameters.
#### - The difference in the standard deviation of per-period log-likelihood between training and test data. 
#### For all these measures, the lower the better.

```python
# Finding the best models

random_states = np.linspace(0, 9999, 50)
n_components = [2, 3, 4, 5, 6]
covariance_types = ["spherical", "diag", "full", "tied"]

models = []

for r in random_states:
    for n in n_components:
        for c in covariance_types:
            model = GaussianHMM(
                n_components = n
                , covariance_type = c
                , n_iter = 1000
                , random_state = int(r))
                    
            model.fit(X_train)
            models.append([
                commodity
                , int(r)
                , n
                , c
                , model.bic(X_train)
                , model.score(X_train) / len(X_train)
                , model.score(X_test) / len(X_test)
                , ((model.score(X_train) / len(X_train)) - (model.score(X_test) / len(X_test)))
            ])

models_df = pd.DataFrame(models, columns = ["commodity", "random_state", "n_components", "covariance_type", "bic", "train_score", "test_score", "train_test_diff"])
models_df["best_random_state"] = models_df.sort_values(by = "bic", ascending = True).groupby(["n_components", "covariance_type"]).cumcount() + 1
```

```python
best_random_states = models_df[models_df["best_random_state"] == 1][["random_state", "n_components", "covariance_type"]].copy()

best_models = (
    models_df
    .groupby(["commodity", "n_components", "covariance_type"]).agg(
        min_bic = ("bic", "min")
        , std_bic = ("bic", "std")
        , std_train_test_diff = ("train_test_diff", "std"))
    .sort_values(by = "min_bic")
    .reset_index(drop = False)
    .merge(best_random_states
           , how = "inner"
           , on = ["n_components", "covariance_type"]))

best_models.head(5)
```

### Best performing models
#### - The top 3 are within 10 BIC, therefore there is no strong evidence for one over another.
#### - Of the top 3, the 2nd and third are more stable over random states.
####
#### We'll try these 3 models on the full dataset.

```python
# Standardising full dataset

scaler = StandardScaler()
X = scaler.fit_transform(df[commodity_cols].copy().to_numpy())
```

```python
# Gathering hyperparameters for best models

count_best_models = 3

best_models_dict = {}
    
for model in range(0, count_best_models):
    best_models_dict[model] = {
        "model_no" : model
        , "n_components" : best_models.iloc[model]["n_components"]
        , "covariance_type" : best_models.iloc[model]["covariance_type"]
        , "random_state" : best_models.iloc[model]["random_state"]}
```

```python
# Testing best models on full dataset

for model_no in range(0, count_best_models):
    model = GaussianHMM(
        n_components = best_models_dict[model_no]["n_components"]
        , covariance_type = best_models_dict[model_no]["covariance_type"]
        , n_iter = 1000
        , random_state = best_models_dict[model_no]["random_state"])

    model.fit(X)
    print(f"{model_no} - {best_models_dict[model_no]['n_components']}, {best_models_dict[model_no]['covariance_type']} - BIC: {np.round(model.bic(X), 2)}, LL: {np.round(model.score(X) / len(X), 2)}")
    best_models_dict[model_no]["model"] = model
    best_models_dict[model_no]["states"] = pd.Series(model.predict(X), name = "states")
    best_models_dict[model_no]["transition_probabilities"] = np.round(model.transmat_, 2)
    best_models_dict[model_no]["bic"] = np.round(model.bic(X), 2)
    best_models_dict[model_no]["log-likelihood"] = np.round(model.score(X) / len(X), 2)
```

#### These results are much more separated with regards to BIC. Whilst models 0 and 2 have the lowest BIC, for a more interesting comparison we'll carry forward the best performing 4 and 5 state models.

### Understanding the states
#### We'll now look at how the states differ by looking at the in state feature means for each model. I have reviewed them and added state_description annotations below.

```python
# Analysing feature column mean values for each state

final_models = [0, 1]

for model_no in final_models:
    # print("")
    # print(f"Model No.{model_no}")
    # print("_______________________")
    states = pd.concat([df[commodity_cols]
                        , best_models_dict[model_no]["states"]]
                        , axis = 1)
    state_means = states.groupby("states")[commodity_cols].mean()
    mean_vals = df[commodity_cols].mean(axis = 0).to_frame(name = "mean").T
    max_vals = df[commodity_cols].max(axis = 0).to_frame(name = "max").T
    min_vals = df[commodity_cols].min(axis = 0).to_frame(name = "min").T
    state_means = pd.concat([state_means, mean_vals, max_vals, min_vals], axis = 0)
    state_means = state_means.rename(columns = {f"{commodity}_price_trend" : "trend", f"{commodity}_return_volatility" : "volatility", f"{commodity}_return_shock" : "shock"}) 
    best_models_dict[model_no]["state_means"] = state_means
    # display(state_means)
```

```python
# Adding state interpretations

state_descriptions = {
    0 : [
        "State 0 - Stable period of growth"
        , "State 1 - Stable period of minor decline"
        , "State 2 - Volatile period of growth"
        , "State 3 - Volatile period of decline"
        , "State 4 - Price shock, High volatility"]
    , 1 : [
        "State 0 - Stable period of growth"
        , "State 1 - Stable period of minor decline"
        , "State 2 - Price shock, High volatility"
        , "State 3 - Volatile period"]
}

for model_no, descriptions in state_descriptions.items():
    print("")
    print(f"Model No.{model_no}")
    print("_______________________")
    state_description = pd.DataFrame(descriptions, columns = ["state_description"])
    best_models_dict[model_no]["state_means"] = pd.concat([best_models_dict[model_no]["state_means"], state_description], axis = 1).fillna("")
    display(best_models_dict[model_no]["state_means"])
```

#### Essentially the 5-state model splits volatility into 2 states: volatile growth and volatile decline.

### Visualising the states
#### Whilst fit metrics and indentification of distinct states is important, we should also check that states temporally align with actual data.

```python
# Visualising states alongside commodity returns

for model_no in final_models:
    print("")
    print(f"Model No.{model_no}")
    [print(x) for x in best_models_dict[model_no]["state_means"]["state_description"] if x != ""]
    print("_______________________")
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    fig.add_trace(
        go.Scatter(
            x = df["date"]
            , y= best_models_dict[model_no]["states"]
            , mode = "lines"
            , name = "state"
            , line = dict(color = "#12436D"))
        , secondary_y = False)
    fig.update_yaxes(showgrid = False, zeroline = False, tickmode = "linear", dtick = 1, secondary_y = False)
    fig.add_trace(
        go.Scatter(
            x= df["date"]
            , y= df[f"{commodity}_return"]
            , mode = "lines"
            , name = "return"
            , line = dict(color = "#F46A25"))
        , secondary_y = True)
    fig.update_yaxes(showgrid = False, secondary_y = True)
    fig.update_layout(height = 200, margin = dict(t = 20, b = 20), plot_bgcolor = "#FAFDFF")
    fig.show()
```

#### Lastly, one should check state transition matrices to understand how state transition behaviour - how likely states are to transition from one to another. A well fit model should demonstrate persistance of states. A poorly fit model exhibits erratic transition behaviour and failure to distinguish states.

```python
# Visualising state transition matrices

for model_no in final_models:
    print("")
    print(f"Model No.{model_no}")
    [print(x) for x in best_models_dict[model_no]["state_means"]["state_description"] if x != ""]
    print("_______________________")
    fig = px.imshow(best_models_dict[model_no]["transition_probabilities"], text_auto = True, color_continuous_scale = "Blues")
    fig.update_xaxes(side="top", tickmode = "linear", dtick = 1, title = dict(text = "future_state"))
    fig.update_yaxes(tickmode = "linear", dtick = 1, title = dict(text = "current_state"))
    fig.update_layout(height = 250, margin = dict(t = 20, b = 20))
    fig.show()
```

#### Both models exhibit a high level of state persistence, an indicator of a well-fit model and realistic state to state transitional probabilities. Models that struggle to separate states often exhibit erratic transitional behaviour.

#### Ultimately, it was felt that the project was a success both, both 4 and 5-state models offer insight into underlying market regimes. Preferance depends on the value one places on separating volatile growth from volatile decline.
